```python
# SVM Classification on Fashion-MNIST
# Linear, Polynomial and RBF Kernels

import pandas as pd
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score


# 1. Load the dataset
train = pd.read_csv("fashion-mnist_train.csv")
test = pd.read_csv("fashion-mnist_test.csv")

print("Training data:", train.shape)
print("Testing data :", test.shape)


# 2. Separate input and output
X_train = train.drop("label", axis=1)
y_train = train["label"]

X_test = test.drop("label", axis=1)
y_test = test["label"]


# 3. Normalize pixel values
X_train = X_train / 255.0
X_test = X_test / 255.0



```

```python
# ---------------------------------------------------------
# 4. Linear Kernel
# ---------------------------------------------------------

linear_model = SVC(kernel="linear")

linear_model.fit(X_train, y_train)

linear_pred = linear_model.predict(X_test)

linear_accuracy = accuracy_score(y_test, linear_pred)

print("Linear Kernel Accuracy:", linear_accuracy)


```

```python
# ---------------------------------------------------------
# 5. Polynomial Kernel
# ---------------------------------------------------------

poly_model = SVC(kernel="poly", degree=3)

poly_model.fit(X_train, y_train)

poly_pred = poly_model.predict(X_test)

poly_accuracy = accuracy_score(y_test, poly_pred)

print("Polynomial Kernel Accuracy:", poly_accuracy)


```

```python
# ---------------------------------------------------------
# 6. RBF Kernel
# ---------------------------------------------------------

rbf_model = SVC(kernel="rbf")

rbf_model.fit(X_train, y_train)

rbf_pred = rbf_model.predict(X_test)

rbf_accuracy = accuracy_score(y_test, rbf_pred)

print("RBF Kernel Accuracy:", rbf_accuracy)

```

```python

# ---------------------------------------------------------
# 7. Compare results
# ---------------------------------------------------------

print("\n--- SVM Kernel Comparison ---")
print("Linear      :", linear_accuracy)
print("Polynomial  :", poly_accuracy)
print("RBF         :", rbf_accuracy)
```

```python
import matplotlib.pyplot as plt

# Accuracy values
kernels = ["Linear", "Polynomial", "RBF"]
accuracy = [linear_accuracy, poly_accuracy, rbf_accuracy]

# Create bar graph
plt.bar(kernels, accuracy, color=["blue", "orange", "green"])

plt.xlabel("SVM Kernel")
plt.ylabel("Accuracy")
plt.title("Comparison of SVM Kernels on Fashion-MNIST")

# Display accuracy on top of each bar
for i in range(len(accuracy)):
    plt.text(i, accuracy[i] + 0.01,
             round(accuracy[i], 3),
             ha="center")

plt.ylim(0, 1)
plt.show()
```
